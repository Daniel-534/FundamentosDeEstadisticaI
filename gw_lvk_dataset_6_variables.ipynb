# Dataset de ondas gravitacionales (6 variables)

Cuaderno breve para obtener un dataset público de eventos de ondas gravitacionales con 6 columnas: 2 cualitativas nominales, 3 cuantitativas continuas y 1 cuantitativa discreta.

## Origen del dataset

La fuente es el catálogo público de eventos del **Gravitational Wave Open Science Center (GWOSC)** / colaboraciones LIGO-Virgo-KAGRA (LVK).

En este cuaderno se usa un respaldo en GitHub con el JSON bruto del catálogo de eventos para facilitar una descarga estable en clase.

## Significado de las 6 variables

1. **evento** (cualitativa nominal): nombre del evento detectado (ej. `GW150914`).
2. **catalogo** (cualitativa nominal): release/catálogo donde fue reportado el evento.
3. **masa_1_fuente_msol** (cuantitativa continua): masa de la componente primaria en masas solares, en el marco fuente.
4. **distancia_luminosidad_mpc** (cuantitativa continua): distancia de luminosidad estimada en megapársecs (Mpc).
5. **snr_red** (cuantitativa continua): relación señal-ruido (SNR) combinada de la red de detectores.
6. **version_evento** (cuantitativa discreta): versión entera del registro del evento (1, 2, 3, ...).

```python
import json
import urllib.request
import pandas as pd
```

```python
# Fuente pública (respaldo del catálogo GWOSC/LVK)
url = 'https://raw.githubusercontent.com/rickyjreyes/GWTC/main/data/gwtc_allevents_raw.json'

with urllib.request.urlopen(url) as response:
    data = json.load(response)

rows = []
for _, e in data['events'].items():
    rows.append({
        'evento': e.get('commonName'),
        'catalogo': e.get('catalog.shortName'),
        'masa_1_fuente_msol': e.get('mass_1_source'),
        'distancia_luminosidad_mpc': e.get('luminosity_distance'),
        'snr_red': e.get('network_matched_filter_snr'),
        'version_evento': e.get('version')
    })

df = pd.DataFrame(rows)

# Mantener solo filas completas y tipos adecuados
df = df.dropna(subset=[
    'evento', 'catalogo', 'masa_1_fuente_msol',
    'distancia_luminosidad_mpc', 'snr_red', 'version_evento'
]).copy()
df['version_evento'] = df['version_evento'].astype(int)

df.head()
```

```python
# Guardar dataset final (6 columnas)
output_file = 'dataset_ondas_gravitacionales_6vars.csv'
df.to_csv(output_file, index=False)
print(f'Se guardó: {output_file} | Filas: {len(df)}')
```
